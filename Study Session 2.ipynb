# APL Study Session 2

[BasedPL](https://github.com/AnswerDotAI/basedpl) (BPL) is an array language derived from APL, with ideas from J and BQN. Functions work on whole arrays, and small functions can be combined into larger ones. Its Rust implementation has a Python interface, so it can run inside SolveIt.

## `x` (Times sign)

### Monadic `x` (Direction)

````
%bpl × 3.1 ¯2 0 3j4
````

For each array element it is divided by it's magnitude to retain a unit value and sign. The complex part is treated exactly the same. If you think of the magnitude as the square root of `9+16` which is `5`. Then we divide `3` and the `4` by this to get `0.6j0.8`, which is a vector in the same direction as before but with unit length.

### Dyadic `x` (Times)

## Precedence

````
%bpl 3×2+1
````

In normal BODMAS evaluation this would result in 7, but in BPL we always evaluate from right to left, and all operators have the same precedence. So we add 1+2 first, *then* multiply by 3.

````
%bpl 1 2 3×2+1
````

````
%bpl 1 2 3×2+2 4 6
````
