## Functions

APL functions operate on whole arrays. A function can be **monadic**, taking a right argument `⍵` when called, or **dyadic**, taking a left argument `⍺` and a right argument `⍵` when called. Direct functions are written inside braces `{…}`.

```
%%bpl
f ← {⍵×2}
f 4
```

- This function is **monadic** because it uses only `⍵`.
- `←` assigns the function to `f`.
- Inside the function, `×2` doubles the supplied value.
- Calling `f 4` therefore returns `8`.

- In a direct monadic function, the supplied argument is referenced with `⍵` (**omega**). It is a fixed symbol, not a parameter name you can replace; using another undefined name results in an error.

A **dyadic** function takes both a left argument `⍺` and a right argument `⍵`. Here, `sum` adds them together.

```
%%bpl
sum ← {⍺+⍵}
3 sum 5
```

A direct function has at most two argument positions, and their symbols are fixed:

- `⍺` (**alpha**) is always the left argument.
- `⍵` (**omega**) is always the right argument.
- Therefore, in `3 sum 5`, `⍺` is `3` and `⍵` is `5`.

Each argument can still be an entire array, so one or two argument positions can carry many values.

In the following two examples this ordering is clear if we swap the function parameter operations. The answer is changed accordingly.

```
%%bpl
g ← {⍺×2+⍵}
3 g 4
```

```
%%bpl
g ← {⍵×2+⍺}
3 g 4
```

In BPL, `←` is the general **assignment** symbol, not specifically a function-assignment symbol. It assigns the value on the right to the target on the left. When the right side is a function, as in `f ← {⍵×2}`, it assigns a function.

In the example below it performs **destructuring assignment** (or unpacking): the bracketed names receive corresponding elements from the array on the right, so `a` becomes `3` and `b` becomes `4`.

```
%%bpl
[a b] ← 3 4
a b
```

We can compose functions too.

```
%%bpl
h ← {2+f⍵}
h 3
```

We can also create anonymous functions and use immediately with a passed in value.

```
%%bpl
{⍵×2} 4
```

## Array Rank

We can assign a rank 1 vector of values to `a`.

```
%%bpl
⎕←a←1 2 3 3 4 5 6
```

`⎕` is pronounced **quad**. It is a special BPL input/output symbol rather than an ordinary variable. When it appears on the left of assignment, `⎕←Y` displays `Y` and also returns `Y`, allowing the value to continue through the surrounding expression.

In `⎕←a←1 2 3`, evaluation proceeds from right to left: the array is first assigned to `a`, then that same array is passed to `⎕←` and displayed.

```
%%bpl
1 + ⎕←2
```

Here, because `⎕←` displays a value without consuming it, it is a convenient debugging tool. You can insert it into a larger expression to inspect an intermediate result while leaving the calculation unchanged.

For example, `1 + ⎕←2` first displays `2`, then passes that value onward so the complete expression still returns `3`.

## ⍴ (Rho)

### Dyadic `⍴` (reshape)

```
%%bpl
⎕←a←1 2 3 4 5 6
```

```
%%bpl
3 2 ⍴ a
```

Reading from right to left, we take the array assigned to `a` and reshape it into **3 rows and 2 columns**.

### Boxed display

Use `•prefs ["box":$t]` to enable boxed display for subsequent results, and `$f` to disable it.

- `→` marks the last axis, running across columns.
- `↓` marks the first axis, running down rows.
- `~` indicates a simple numeric array.

The final example below re-enables boxing for subsequent examples.

```
%%bpl
•prefs ["box":$t]
3 2 ⍴ a
```

```
%%bpl
2 3 ⍴ a
```

Turn boxing off with `$f`; the same reshape is then displayed plainly.

```
%%bpl
•prefs ["box":$f]
3 2 ⍴ a
```

```
%%bpl
•prefs ["box":$t]
3 2 ⍴ a
```

Boxing is now enabled for the remaining examples.

```
%%bpl
⎕←a←1 2 3 4 5 6
```

Now we can reshape it to be a 1x6 matrix.

```
%%bpl
1 6 ⍴ a
```

### Monadic `⍴` (shape of)

Let's first use the dyadic version of reshape to reshape the 1x6 matrix `a` into a 3x2 matrix called `mat`.

```
%%bpl
⎕←mat←2 3 ⍴ a
```

Now we can just use the monadic reshape to return the shape of `a`.

```
%%bpl
⍴a
```

Which is a single dimension of length 6. And for `mat` the shape is:

```
%%bpl
⍴mat
```

2 rows and three columns.

```
%%bpl
⍴3
```

This is Zilde the empty vector in APL, as shown below too.

```
%%bpl
⍬
```

```
%%bpl
2 2 3 ⍴ a
```

```
%%bpl
2 3 ⍴ 1
```
