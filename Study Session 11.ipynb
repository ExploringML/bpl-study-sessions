## Rank and Reduction

Adám Brudzewsky describes the rank operator `⍤` as putting 'blinkers' on a function: it controls how much of an array the function sees in each call. An array's rank is its number of axes. Here, rank 2 lets the function see one matrix at a time. We can reproduce the screenshot in BPL using `⌿` (reduce first) and `⍤` (rank).

````
%%bpl
⎕ ← b ← ↦24
⎕ ← a ← 2 3 4 ⍴ ↦24
````

`↦24` generates 1 through 24. Reshape (`⍴`) arranges them into `a` with shape `2 3 4`: two planes, each containing three rows of four numbers. The first plane holds 1–12 and the second holds 13–24. There are three axes, so `a` has rank 3.

````
%%bpl
+⌿a
````

`+⌿` adds along the first axis of the whole array. That axis selects the two planes, so corresponding positions are added: `1+13=14`, `2+14=16`, and so on. The result has shape `3 4`. We have combined the two planes into one matrix.

````
%%bpl
(+⌿⍤2)a
````

`⍤2` puts the blinkers on `+⌿`: each call receives one rank-2 cell, here a `3 4` matrix. Within that matrix, the first axis selects rows, so `+⌿` adds the three rows together: `1+5+9=15` in the first plane and `13+17+21=51` in the second. Rank assembles the two four-number results into a `2 4` matrix. The same reduction now works within each plane. Narrowing the view further to rank 1 would let it see just one row per call and sum that row, as below.

````
%%bpl
(+⌿⍤1)a
````

`⍤1` makes six calls to `+⌿`, each receiving one row of four numbers and returning its sum. Each of the two planes contributes three row sums, so the result has shape `2 3`: `10 26 42` from the first plane and `58 74 90` from the second.

Split the original shape as `2 3 | 4`: `4` is the shape each call sees; `2 3` is the **frame**, which records where each row came from. Replacing each row with one scalar preserves that frame.

## Averages and the Leading Axis

To calculate an average, the sum and the count must describe the same collection of values. 'Leading axis' means the first axis of the argument currently received by the function. Rank can change that argument by supplying smaller cells.

````
%%bpl
m←2 3⍴↦6
m
⍝ This version mixes the last-axis sum with a first-axis count.
{(+/⍵)÷≢⍵}m
````

`m` has two rows and three columns. `+/m` sums along the last axis, giving the row sums `6 15`. But `≢m` counts along the first axis: it returns the number of rows, `2`. The result is therefore `6÷2` and `15÷2`, or `3 7.5`, rather than the row averages. The sum and count refer to different axes.

````
%%bpl
avg←{(+⌿⍵)÷≢⍵}
avg m
````

`+⌿` reduces the first (leading) axis, adding the two rows together to give `5 7 9`. Tally (`≢`) counts those same two rows. Dividing by `2` gives the column averages `2.5 3.5 4.5`. This definition keeps the sum and the count aligned along the same axis.

````
%%bpl
(avg⍤1)m
````

Rank 1 supplies one row at a time to the same `avg` function. For `1 2 3`, the sum is `6` and tally is `3`, giving `2`; for `4 5 6`, they give `15÷3=5`. The result is `2 5`. Each row is a vector, so its only axis is now its leading axis. This is the advantage of the first-axis definition: rank chooses the cells, while the averaging function keeps summing and counting the same collection.

## Reverse, Reverse First and Transpose

`⌽` (reverse), `⊖` (reverse first) and `⍉` (transpose) are functions: each acts directly on an array. Rank (`⍤`) is an operator that takes a function and controls which cells it receives. On a matrix, the lines through these three circles suggest vertical, horizontal and diagonal mirror lines. Use the same `2 3` matrix as in the averages examples.

````
%%bpl
m←2 3⍴↦6
m
⌽m
````

`⌽` reverses the last axis. On this matrix, it reverses the numbers within each row: `1 2 3` becomes `3 2 1`, and `4 5 6` becomes `6 5 4`. Think of a left-to-right reflection across the vertical line in the glyph. The shape stays `2 3`.

````
%%bpl
⊖m
````

`⊖` reverses the first (leading) axis. Here that swaps the two whole rows: `4 5 6` comes first, followed by `1 2 3`. The horizontal bar suggests a top-to-bottom reflection. The shape stays `2 3`. This first-axis/last-axis distinction is the same one we saw with `+⌿` and `+/`.

````
%%bpl
m
⍉m
````

`⍉` transposes the matrix, exchanging its row and column axes. The original columns become rows: `1 4`, `2 5` and `3 6`. Its shape changes from `2 3` to `3 2`. Think of reflection across the top-left to bottom-right diagonal. More generally, monadic transpose reverses the order of an array's axes.

````
%%bpl
m
(⌽⍤1)m
(⊖⍤1)m
````

The cell prints `m` first, then tries two ways of reversing its rows: `(⌽⍤1)m` and `(⊖⍤1)m`. The two matching outputs are expected.

Both expressions contain `⍤1`. This tells rank to give the function one rank-1 cell at a time. For our matrix, these are the vectors `1 2 3` and `4 5 6`, taken from the last axis. Each call receives just those three numbers. A vector has only one axis, so its first axis and its last axis are the same axis.

1. In `(⌽⍤1)m`, `⌽` means reverse the last axis. Given `1 2 3`, it returns `3 2 1`. Given `4 5 6`, it returns `6 5 4`.
2. In `(⊖⍤1)m`, `⊖` means reverse the first axis. It receives the same two vectors. Each vector has only one axis, so it reverses those same numbers: `1 2 3` becomes `3 2 1`, and `4 5 6` becomes `6 5 4`.

Rank puts the answers back in their original row positions. Both expressions therefore produce:

```text
3 2 1
6 5 4
```

The distinction matters when `⊖` receives the whole matrix. `⊖m` sees a first axis containing two whole rows, so it swaps those rows. In `(⊖⍤1)m`, rank supplies a vector to each call, and that vector's first axis contains three numbers, so those numbers are reversed instead.

For this horizontal reversal, `⌽m` already does the job directly; adding `⍤1` to `⌽` gives the same result here. The ranked `⊖` example demonstrates the same useful idea as `(avg⍤1)m`: a function that works along the first axis can be made to work within individual rows by controlling what it receives.

````
%%bpl
m
⍉m
(avg⍤1)⍉m
````

Transpose first makes the original columns into rows. Rank 1 then gives `avg` those vectors, producing `2.5 3.5 4.5`, the original column averages. Rank always selects cells from the trailing (last) axes; transpose changes which original axis occupies that position. Reversal changes the order of items along an axis, while transpose changes the order of the axes themselves.

## Equality, Match and Rank 0

The screenshot compares the vectors `3 4` and `3 5`. `=` is equality: it compares individual values. `≡` (three horizontal bars) is match: it checks whether two whole arrays have the same shape and matching contents. It is a different glyph from tally, `≢`, which we used for averages. In the video, `1` means true and `0` means false; BPL displays these Boolean results as `$t` and `$f`.

````
%%bpl
3 "D" "Q" 4=3 "D" "W" 5
````

Equality compares corresponding numbers: `3=3` is true, and `4=5` is false. There are two comparisons, so the result is the two-item vector `$t $f`, corresponding to the video's `1 0`. Equality is a scalar function: it does its work on individual values even when supplied with arrays.

````
%%bpl
3 4≡3 5
````

Match asks one question: "Do these two complete vectors match?" Their shapes agree, but the second values differ (`4` and `5`), so the answer is one scalar, `$f`. The first pair matching is not enough for the whole vectors to match.

````
%%bpl
3 4≡3 5
3 4≡3 5 2
3 5 2≡3 5 2
3 4(≡⍤0)3 5
````

`≡` normally compares two complete arrays and returns one Boolean:

- `3 4≡3 5` gives `$f`: the vectors have the same shape, but `4` and `5` differ.
- `3 4≡3 5 2` also gives `$f`: their lengths, and therefore their shapes, differ. Match can compare different shapes; they simply do not match.

In `3 4(≡⍤0)3 5`, rank 0 gives `≡` one pair of individual values per call. It matches `3` with `3` (true), then `4` with `5` (false), and collects the answers into `$t $f`. Each call still compares its two complete inputs, but those inputs are now single values.

Rank 0 means a scalar with no axes; the `0` is not an axis number. This is the same "blinkers" idea as giving `avg` one row at rank 1, narrowed here to one value at rank 0.

````
%%bpl
3 4(=⍤1)3 5
3 4(≡⍤1)3 5
````

At rank 1, each function receives the two whole vectors. Equality still compares their numbers separately and returns `$t $f`; match compares the whole vectors and returns `$f`. Rank controls the inputs passed to a function. It does not change what that function does with those inputs. This is why increasing the rank of equality does not make it behave like whole-array match.

````
%%bpl
3 4=3 5
3 4(≡⍤0)3 5
````

`3 4=3 5` and `3 4(≡⍤0)3 5` both give `$t $f`. Equality compares corresponding numbers directly. Rank 0 makes match receive `3` and `3` in its first call, then `4` and `5` in its second, giving true and false. It never compares the two whole vectors in one call. For these numbers, match at rank 0 therefore gives the same result as ordinary equality.

## Debugging by Printing Function Arguments

To see what a function receives, add a printing statement before its calculation. In `{⎕←⍺ ⍵ ⋄ ⍺=⍵}`, `⍺` (alpha) is the left argument and `⍵` (omega) is the right argument.

The diamond `⋄` separates two statements, which run in the order written:

1. `⎕←⍺ ⍵` groups the two arguments and prints them.
2. `⍺=⍵` compares the original arguments and supplies the function's return value.

BPL evaluates expressions from right to left, but that does not reverse the order of statements separated by `⋄`. It finishes the printing statement before starting the equality statement. SolveIt then displays the returned answer separately. BPL uses `$t $f` for the video's `1 0`.

````
%%bpl
3 4{⎕←⍺ ⍵ ⋄ ⍺=⍵}3 5
````

Here the function runs once, with the whole vector `3 4` as `⍺` and the whole vector `3 5` as `⍵`. Inside the braces:

1. `⍺ ⍵` forms a two-item **nested vector**: its first item is the vector `3 4`; its second is the vector `3 5`. The space between the arguments groups them as items; it does not flatten them into the four-number vector `3 4 3 5`.
2. `⎕←` prints that nested vector. The curved `∊` on the bottom edge of the outer box marks nested storage: this array contains other arrays. It is a display label here, not an operation being executed. The right-pointing arrows mark the outer vector and its two inner vectors as one-dimensional.
3. At `⋄`, BPL moves to the next statement, `⍺=⍵`. The arguments are still `3 4` and `3 5`. Equality compares corresponding numbers: `3=3` gives `$t`, and `4=5` gives `$f`.
4. This last statement returns the Boolean vector `$t $f`, which SolveIt displays below the printed inputs.

The first output therefore shows **what went into the function**; the second shows **what the calculation returned**.

````
%%bpl
3 4({⎕←⍺ ⍵ ⋄ ⍺=⍵}⍤0)3 5
````

Here `⍤0` applies to the entire function in braces, including its print statement. Rank calls that wrapper twice, each time supplying one value on each side:

| Call | Inputs printed | Comparison returned |
|---|---|---|
| First | `3 3` | `$t` (true) |
| Second | `4 5` | `$f` (false) |

The output therefore shows `3 3`, then `4 5`, then the final vector `$t $f` assembled by rank. The first two outputs reveal the inputs; the last is the answer.

Both examples use the same equality calculation and give the same answer. What changes is the wrapper's inputs and number of calls: once with two vectors, or twice with pairs of numbers. The two comparisons are `3` with `3`, then `4` with `5`. This demonstrates that rank controls what the whole function receives; it does not change the calculation inside it.
