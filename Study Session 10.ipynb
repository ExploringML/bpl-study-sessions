## Attack of the Mutations

The puzzle asks how many corresponding positions differ between two equal-length DNA strings. This count is the **Hamming distance**. We first counts matches, then change the function to count differences.

For Atop, BPL uses `∘` where the screenshot uses APL `⍤`. BPL also uses double quotes for strings. [BPL Atop reference](https://answerdotai.github.io/basedpl/glyphs/jot.html).

### Counting Matching Positions

```
%%bpl
leftDNA ← "GAGCCTACTAACGGGAT"
rightDNA ← "CATCGTAATGACGGCCT"
```

```
%%bpl
f ← =
leftDNA f rightDNA
```

```
%%bpl
f ← +/∘=
leftDNA f rightDNA
```

Atop not strictly necessary here.

```
%%bpl
f ← +/=
leftDNA f rightDNA
```

1. `/` combines with `+` to create the summing function `+/`.
2. Atop `∘` combines `+/` and `=` to create `f`. When we call `leftDNA f rightDNA`, **both strings go to `=` first**. It compares the first character with the first, the second with the second, and so on.
3. `=` returns a true/false value for each position. `+/` sums that vector, counting true as `1` and false as `0`. The result is **10 matching positions** out of 17.

Read `+/∘=` as "compare for equality, then sum the results".

### Counting Different Positions

```
%%bpl
⍝ Tally counts the items in each DNA vector
≢ leftDNA
≢ rightDNA
```

`≢` ([Tally](https://answerdotai.github.io/basedpl/glyphs/equal-underbar-slash.html)) is a function: with one vector, it counts its items. Both DNA strings contain 17 characters. It is a different glyph from `≠` (Not Equal).

```
%%bpl
⍝ 1. Compare corresponding characters
leftDNA = rightDNA

⍝ 2. Flip true and false
~ (leftDNA = rightDNA)

⍝ 3. Count the differences
+/ ~ (leftDNA = rightDNA)
```

1. `=` produces the equality mask: `$t` for a match, `$f` for a difference.
2. `~` means **NOT**. It flips every value, so a difference now gives `$t` and a match gives `$f`. For example, the first characters, `G` and `C`, differ: false becomes true. The second characters are both `A`: true becomes false.
3. `+/` sums the flipped mask, giving **7 different positions**. That is the Hamming distance: 17 positions = 10 matches + 7 differences.

```
%%bpl
matches ← leftDNA = rightDNA
differences ← ~matches

⍝ Count the true values: matching positions
+/ matches

⍝ Count the false values by flipping them to true: different positions
+/ differences
```

The equality mask has 10 true values and 7 false values. `+/ matches` counts the true values; `+/ differences` counts the original false values, because `~` has flipped them to true.

```
%%bpl
⍝ Collect letters from the matching positions. Can be either leftDNA or rightDNA as the matches are common to both original strings
⎕ ← matchingDNA ← matches # leftDNA

⍝ Collect letters from the differing positions in each string
differentLeftDNA ← differences # leftDNA
differentRightDNA ← differences # rightDNA
differentLeftDNA
differentRightDNA
```

`#` ([Replicate](https://answerdotai.github.io/basedpl/glyphs/hash.html)) keeps a letter where its mask is `$t` and drops it where the mask is `$f`. `matchingDNA` contains the 10 matching letters (the same in both strings). `differentLeftDNA` and `differentRightDNA` contain the 7 letters from the differing positions, in their original order.

```
%%bpl
f ← +/~∘=
leftDNA f rightDNA
```

This redefines `f` using a modified expression. Read `+/~∘=` as **compare → NOT → sum**.

- `+/~` is a two-function train: it applies `~` first, then `+/`.
- `∘` combines that train with `=`. Both string arguments go to `=`; its result then goes through `~` and `+/`.
- The whole call is equivalent to `+/ (~ (leftDNA = rightDNA))`, which gives `7`.

The extra `~` changes what we count: true now marks a difference.

### Using Not Equal Directly

```
%%bpl
⍝ Not equal already marks differences as true
leftDNA (+/∘≠) rightDNA

⍝ Other puzzle examples, using a modified f
"A" f "T"
"" f ""
rightDNA f rightDNA
```

With two inputs, `≠` means "not equal", so `+/∘≠` compares for differences and sums them directly: the same result, `7`.

For the remaining examples, BPL displays one difference as `$t` (true, numerically `1`). Empty strings have no differences, and comparing a string with itself also gives `0`.

## Quad Equal (Key): Grouping Cards by Suit

`⌸` (Key) groups items that have the same key, then applies a function to each group. BPL uses the same glyph as APL.

```
%%bpl
⎕ ← a ← "banana"
```

This stores the six-character vector `"banana"` in `a`.

```
%%bpl
⎕ ← cards ← ["2" "Queen" "Ace" "4" "Jack"]
⎕ ← suits ← ["Spades" "Hearts" "Spades" "Clubs" "Hearts"]
```

The vectors line up by position: `"2"` is a Spade, `"Queen"` a Heart, and so on. BPL uses brackets for the vector of strings and double quotes for each string.

```
%%bpl
suits {[⍺ ":" ⍵]} cards
```

Without `⌸`, the function is called **once**: `⍺` receives the whole `suits` vector and `⍵` the whole `cards` vector. `[⍺ ":" ⍵]` simply returns those two vectors with `":"` between them.

Adding `⌸` below changes which arguments the function receives, and how many times it is called.

```
%%bpl
suits {[⍺ ":" ⍵]}⌸ cards
```

`⌸` (Key) is an **operator** applied to `{[⍺ ":" ⍵]}`. Here is how it prepares the arguments for that function:

1. **Keep the positions linked:** position 0 in `suits` labels position 0 in `cards`, position 1 labels position 1, and so on. BPL positions start at 0.
2. **Find the positions for each distinct suit:** Spades → `0 2`; Hearts → `1 4`; Clubs → `3`.
3. **Take cards at those same positions:** Spades → `["2" "Ace"]`; Hearts → `["Queen" "Jack"]`; Clubs → `["4"]`. These are the grouped card arguments that Key prepares; you do not need to save them in named variables.
4. **Call your function for each group:** pass the suit as `⍺` and that group's card vector as `⍵`. The body returns `[⍺ ":" ⍵]`.
5. **Assemble the returned rows** into the displayed result.

For these inputs, the three calls receive exactly the arguments shown below:

```
%%bpl
"Spades" {[⍺ ":" ⍵]} ["2" "Ace"]
"Hearts" {[⍺ ":" ⍵]} ["Queen" "Jack"]
"Clubs" {[⍺ ":" ⍵]} ["4"]
```

The three calls above show what **your function receives** after grouping. The function body only builds a row; it does not find matching suits or select cards. **`⌸` handles that work:** it uses matching positions in the two input vectors to prepare each suit/card group, calls your function with each pair, and assembles the returned rows. The notebook displays the result.

Here are some further examples:

```
%%bpl
a {[⍺ ⍵]}⌸ 3 1 4 1 5 9
```

```
%%bpl
a {[⍺ ⍵]}⌸ ↦6
⍴(a {[⍺ ⍵]}⌸ ↦6)
```

```
%%bpl
a ,⌸ ↦6
⍴(a ,⌸ ↦6)
```

```
%%bpl
a {⍺, +/⍵}⌸ ↦6
```

```
%%bpl
a {⍺ ⍵}⌸ ↦6
a {⍵}⌸ ↦6
a {+/⍵}⌸ ↦6
```

```
%%bpl
,⌸ a
```

```
%%bpl
{≢ ⍵}⌸ a
```

```
%%bpl
{+/⍵}⌸ a
```

With just one input, `,⌸ a` uses the letters in `"banana"` as keys and automatically supplies their **zero-based positions** as values. It groups these positions by letter: `b → 0`, `a → 1 3 5`, `n → 2 4`. For each group, `⌸` calls comma with the letter on the left and its position vector on the right; comma joins them into the displayed row.

This is the **monadic call of the function created by `,⌸`**. Comma still receives two arguments for each group: the letter and its positions.

## Equal Underbar Slash: Tally and Not Match

`≢` is one glyph. With one argument it means **Tally** (count items); with two it means **Not Match** (compare the whole values, including their shapes).

```
%%bpl
≢ "banana"
```

Tally counts the six characters, giving `6ₓ` (exactly 6).

```
%%bpl
1 ≢ 1
```

The two values match, so Not Match returns `$f` (false), equivalent to the screenshot's `0`.

```
%%bpl
1 ≢ 0
```

The values differ, so the result is `$t` (true), equivalent to `1`.

```
%%bpl
1 ≢ 1 1
```

A single number and a two-item vector have different shapes, so they do not match: `$t`. Unlike `≠`, `≢` compares the whole inputs and returns one Boolean result.

## Outer Product: Joining Every Pair

An **outer product** applies a function to every possible pair of items from two inputs. Jeremy uses APL `∘.,`; the BPL equivalent is `,⊗`, where `⊗` modifies comma (Catenate).

```
%%bpl
a ← "AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC"
```

`a` now holds this 70-letter DNA sequence, replacing the earlier `"banana"` value.

```
%%bpl
c ← "ACGT"
```

`c` contains the four DNA bases. Each one will be paired with every character in `a`.

```
%%bpl
c ,⊗ a
```

`,⊗` calls comma for every pair: `A` with `A` gives `AA`, and `A` with `G` gives `AG`. The four rows join `A`, `C`, `G` and `T` respectively to every character in `a`, producing a **4-by-70 array of two-letter strings**.

```
%%bpl
c =⊗ a
```

`=⊗` uses the same pairs but **compares the letters instead of joining them**. BPL displays `$t` for a match and `$f` for no match, corresponding to Jeremy's `1` and `0`. The four rows mark the positions of `A`, `C`, `G` and `T` in `a`. This is a **Boolean mask**: it marks matches and keeps every position.

```
%%bpl
+/ c =⊗ a
```

`+/` sums each row of the Boolean grid, counting true as `1` and false as `0`. The result `20 12 17 21` gives the counts of **A, C, G and T**, in the order used by `c`.
