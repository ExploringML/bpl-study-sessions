## Python Decorators Review

```
def add1(x): return x+1
```

```
add1
```

```
add1(3)
```

```
def logargs(f):
    return f
```

```
g = logargs(add1)
g
```

```
def logargs(f):
    def _inner(*args, **kwargs):
        print(f"Received args: {args}; kwargs: {kwargs}")
        return f(*args, **kwargs)
    return _inner
```

```
g = logargs(add1)
g
```

```
g(3)
```

We can do the same with decorators, and pass the `add1` function to `logargs` when we define it, as follows:

```
@logargs
def add1(x): return x+1
```

When Python defines `add1`, the decorator runs once: `add1 = logargs(add1)`. The name `add1` then refers to the `_inner` wrapper. Later, calling `add1(5)` calls `_inner(5)`, which logs the arguments and calls the original `add1` function.

```
add1(5)
```

```
@logargs
def mult2(x): return x*2
```

```
mult2(7)
```

## BPL Operators

In BPL functions that take in a function and return a function are called operators.

### `/` (Slash)

#### monadic `/` (Reduce)

```
%%bpl
⎕ ← a ← ⍳5
+/a
```

```
%%bpl
⎕ ← a ← 1 + ⍳5
×/a
```

Read `+/a` as **“plus reduce `a`”**, or **“sum `a`.”** The `/` operator combines all elements using `+`, so for `a ← 0 1 2 3 4`, the result is `0 + 1 + 2 + 3 + 4 = 10`.

```
%%bpl
a ← 4 6 2
⌈/a
```

This will insert the upstile glyph between the elements of `a` and evaluate from right to left, like `4 ⌈ 6 ⌈ 2`.

### `\` (Slash)

#### monadic `\` (Reduce)

```
%%bpl
⎕ ← a ← ⍳5
+\a
```

```
%%bpl
⎕ ← a ← 1 + ⍳5
×\a
```

## Min and Max

### `⌈` (Upstile)

#### monadic `⌈` (Ceiling)

```
%%bpl
⌈ 3.4 ¯3.4 3 0
```

#### dyadic `⌈` (Maximum)

```
%%bpl
3⌈2
```

```
%%bpl
3 2⌈2 3
```

```
%%bpl
a ← 4 6 2
4⌈6⌈2
```

#### ReLU in BPL

With two inputs (dyadic), `⌈` (upstile) returns the larger value for each element. In `0 ⌈ values`, negative values become zero, while zero and positive values stay unchanged. This is **ReLU** (rectified linear unit): `max(0, x)`. Zero sets a **lower bound**; the list keeps the same length and order.

```
%%bpl
values ← ¯4 0 6 2
0 ⌈ values    ⍝ ReLU: 0 0 6 2
```

### `⌊` (Downstile)

#### monadic `⌊` (Floor)

```
%%bpl
⌊ 3.4 ¯3.4 3 0
```

#### dyadic `⌊` (Minimum)

```
%%bpl
3⌊2
```

```
%%bpl
4⌊6⌊2
```
