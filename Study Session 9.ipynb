## ⍤ (Jot Diaeresis)

### Rank and Shape

An array's **shape** gives the length of each dimension. Its **rank** is how many dimensions it has. A vector has rank 1; a matrix has rank 2, however many rows and columns it contains.

In the examples below, `⍴` asks for the shape. `≢` counts items, so `≢ (⍴ array)` counts the entries in the shape: that is the rank. `⎕ ←` displays a value while assigning it, and BPL's `↦6` generates `1 2 3 4 5 6`.

```
%%bpl
⍝ A scalar has no dimensions: rank 0
⍴ 7
≢ (⍴ 7)

⍝ A vector of six numbers: shape 6, rank 1
⎕ ← rank_vector ← ↦6
⍴ rank_vector
≢ (⍴ rank_vector)
```

```
%%bpl
⍝ Two rows, three columns: shape 2 3, rank 2
⎕ ← rank_wide ← 2 3 ⍴ ↦6
⍴ rank_wide
≢ (⍴ rank_wide)

⍝ Three rows, two columns: shape 3 2, still rank 2
⎕ ← rank_tall ← 3 2 ⍴ ↦6
⍴ rank_tall
≢ (⍴ rank_tall)
```

```
%%bpl
⍝ Two layers, each with two rows of three: rank 3
⎕ ← rank_block ← 2 2 3 ⍴ ↦12
⍴ rank_block
≢ (⍴ rank_block)
```

The scalar's shape is `⍬`, an empty vector: it needs no dimensions. A six-item vector has shape `6`, so rank 1. Both matrices have two entries in their shapes, so rank 2. Even a `1 6` matrix has rank 2: a dimension of length 1 still counts. The last array has shape `2 2 3`, so rank 3.

Now separate two questions:

- **Rank: how many dimensions does a piece have?** `f⍤1` applies `f` separately to rank-1 pieces, called cells. These use the last dimension, so they are rows in a matrix. `f⍤2` uses two-dimensional pieces: the whole matrix, or each layer of `rank_block`. `f⍤0` uses individual scalar elements.
- **Axis: which dimension are we working along?** BPL numbers axes from 0. In a matrix of shape `2 3`, axis 0 is the two row positions (down the display); axis 1 is the three column positions (across). An axis number selects a dimension; it does not say how many dimensions a piece contains.

The full glyph `⍤` is called **Jot Diaeresis**, and here it means Rank. The two dots alone, `¨`, are the different operator Each. BPL uses `⍠` for Axis; it is separate from `⍤`.

### Monadic ⍤ (Rank)

```
%%bpl
⎕ ← mat ← 2 3 ⍴ ↦6
(+/⍤1) mat
```

```
%%bpl
mat ← 2 3 ⍴ ↦6
+/ mat
```

```
%%bpl
+⌿mat
```

Read `(+/⍤1) mat` in two stages: first build a function, then give it `mat`.

1. `/` is the Reduce operator. It takes `+` (addition) and creates the function `+/`, which sums along the last axis of its input.
2. `⍤` takes that function `+/` and the rank `1`, creating another function: `+/⍤1`. It says "give the sum function each one-dimensional cell separately".
3. `mat` has shape `2 3`. Its rank-1 cells use the last dimension: the rows `1 2 3` and `4 5 6`.
4. Sum the first row: `1+2+3 = 6`. Sum the second: `4+5+6 = 15`. Assemble those two results into the vector `6 15`.

Ordinary `+/mat` also gives `6 15`, because its default is to reduce along the matrix's last axis. This example shows what Rank means even though ordinary reduction already gives the same answer.

`+⌿` also contains two glyphs: `+` and `⌿`. The single glyph `⌿` (Slash Bar) is the Reduce First operator; it takes addition and creates a function that sums along the first axis. So `+⌿mat` adds down each column: `1+4`, `2+5`, `3+6`, giving `5 7 9`. Both `/` and `⌿` are operators; both `+/` and `+⌿` are the functions they create.

"Monadic" in the subheading describes the call: the resulting function receives one argument, `mat`. The Rank operator `⍤` itself takes two operands, `+/` and `1`, so it is a dyadic operator. [BPL Rank reference](https://answerdotai.github.io/basedpl/glyphs/jot-diaeresis.html).

### Rank Compared with Axis

```
%%bpl
mat ← 2 3 ⍴ ↦6

⍝ Rank 1: sum each row (a one-dimensional cell)
(+/⍤1) mat

⍝ Axis 1: sum across the three column positions in each row
(+/⍠1) mat

⍝ Axis 0: sum down the two row positions in each column
(+/⍠0) mat

⍝ Reduce First is a shorter way to express this last sum
+⌿ mat
```

The first two calls give `6 15`: each row collapses to one sum. The last two give `5 7 9`: each column collapses to one sum.

`⍤1` means "use cells with one dimension", always taking that dimension from the end of the shape. For this rank-2 matrix, the last axis is axis 1, so `⍤1` and `⍠1` select the same pieces. For our rank-3 array `rank_block`, the last axis is axis 2, so rank 1 corresponds to axis 2 instead. **The rank number counts dimensions; the axis number identifies a dimension.**

Rank `⍤` works this way in both APL and BPL. Jeremy's APL notation for an explicit axis is commonly `f[axis]`; BPL uses `f⍠axis`. Also check the numbering: Jeremy's examples use axes starting at 1, whereas BPL axes start at 0. [BPL Axis reference](https://answerdotai.github.io/basedpl/glyphs/quad-colon.html).

### Rank on a Cube

```
%%bpl
⎕ ← cube ← 2 2 2 ⍴ ↦8
(+/⍤1) cube
```

```
%%bpl
(+/⍤2) cube
```

Your reading of the rank is right: `⍤2` gives the function each whole `2 2` matrix. But it does not change what that function does. **`+/` sums along one axis, even when its input is a matrix.**

- With `⍤1`, `+/` receives four vectors: `1 2`, `3 4`, `5 6`, `7 8`. Their sums are `3`, `7`, `11`, `15`. The remaining two dimensions arrange those results into a `2 2` matrix.
- With `⍤2`, `+/` receives two matrices. On the first, it sums the rows to produce `3 7`; on the second, it produces `11 15`. Each result is a vector, so combining the two results again gives the same `2 2` matrix.

To get one total per matrix, the function must sum **all** its elements. We can first ravel each matrix into a vector with monadic comma `,`, then sum that vector.

```
%%bpl
(+/,⍤2) cube
```

Adding ravel `,` gives `(+/,⍤2) cube`. For each rank-2 matrix, `,` turns it into a vector, then `+/` sums that vector. The two-function train `+/,` already means "ravel, then sum", so no explicit `∘` or function body is needed. Result: `10 26`.

### Dyadic ⍤ (Rank)

```
%%bpl
⎕ ← mat ← 3 4 ⍴ ↦12
1 2 3 (+⍤0 1) mat
```

Read `+⍤0 1` as "add a scalar from the left to a vector from the right". Rank `0` takes the individual numbers `1`, `2`, `3`; rank `1` takes the three rows of `mat`. Pair them in order: add `1` to every element of the first row, `2` to the second row, and `3` to the third. These are left/right cell ranks, not axis numbers.

### Atop: APL ⍤, BPL ∘

```
%%bpl
⍝ APL uses *⍤÷ here; BPL uses Atop ∘
f ← *∘÷
```

`f ← *∘÷` creates a function that applies `÷` first, then applies exponential `*` to the result.

```
%%bpl
* (÷3)
```

`÷3` gives the reciprocal `1÷3`; monadic `*` then computes e to that power, giving approximately `1.39561`.

```
%%bpl
f 3
```

```
%%bpl
2 f 3
```

```
%%bpl
*2÷3
```

```
%%bpl
2*(÷3)
```

Atop applies the right function, then passes its result to the left function. APL uses `⍤` for this function-function combination; BPL uses `∘` and keeps `⍤` for Rank.

- `f 3` means `* (÷3)`: reciprocal of 3, then exponential, giving `1.395612425`.
- `2 f 3` means `* (2 ÷ 3)`: divide 2 by 3, then exponential, giving `1.947734041`. Both arguments go to `÷`; `*` receives its single result.
- `2 * (÷3)` instead uses dyadic `*` (power): 2 to the power of one-third, giving `1.25992105`.

Atop is a dyadic operator because it takes two function operands, here `*` and `÷`. The function it creates can be called with one or two arguments. [BPL Atop reference](https://answerdotai.github.io/basedpl/glyphs/jot.html).

### Three Functions: Fork or Pipeline

```
%%bpl
⍝ Fork: first negate the input and take its exponential separately; then divide those two results
(- ÷ *) 2
```

```
%%bpl
⍝ Pipeline: exponential, then reciprocal, then negate
(-∘÷∘*) 2
```

Two adjacent functions form an atop (a pipeline), so an explicit `∘` is optional there. Three plain functions form a fork; use `∘` between stages to request a pipeline instead. Both arrangements are function trains.

- **Fork `(- ÷ *) 2`:** both outside functions receive `2`. Negation gives `¯2`, and exponential gives `*2 ≈ 7.389`. The middle function divides those results: `¯2 ÷ 7.389 ≈ ¯0.27067`.
- **Pipeline `(-∘÷∘*) 2`:** exponential first gives `7.389`, reciprocal gives `0.135335`, then negation gives `¯0.135335`. This reads as `- (÷ (*2))`.

The middle `÷` receives two results in the fork, so it divides; it receives one result in the pipeline, so it takes the reciprocal. [BPL train rules](https://answerdotai.github.io/basedpl/evaluation.html#trains).

## ⍥ (Circle Diaeresis)

### Dyadic ⍥ (Over)

Over `⍥` works the same here in APL and BPL: apply the right function to each argument separately, then apply the left function to the result(s). With one argument it behaves like Atop; with two it transforms both inputs before combining them. [BPL Over reference](https://answerdotai.github.io/basedpl/glyphs/circle-diaeresis.html).

```
%%bpl
f ← *⍥÷
```

`f` first takes reciprocal(s) with `÷`, then applies `*`: exponential for one result, or power for two.

```
%%bpl
* (÷3)
```

`÷3` gives `1÷3`; exponential `*` then computes e to that power, giving approximately `1.39561`.

```
%%bpl
f 3
```

`f 3` performs the same calculation as `* (÷3)`, so it gives the same result.

```
%%bpl
2 f 3
```

Take both reciprocals first: `÷2` gives `1÷2`, and `÷3` gives `1÷3`; then dyadic `*` raises the first result to the second, giving approximately `0.793701`.

```
%%bpl
(÷2)*÷3
```

This writes out `2 f 3`: one-half raised to the power of one-third, confirming the same result.

## # (Replicate)

### Unique Mask and Replicate

Unique Mask `≠` marks the first occurrence of each value with `1` and later copies with `0`. Replicate uses that mask to keep or drop items. APL uses `/` here; BPL uses the dyadic function `#` (mask on the left, values on the right). BPL's `/` is Reduce. [BPL Replicate reference](https://answerdotai.github.io/basedpl/glyphs/hash.html).

```
%%bpl
v ← 22 10 22 22 21 10 5 10
≠v
```

BPL displays mask values as `$t` (true, or 1) and `$f` (false, or 0): the first `22`, `10`, `21` and `5` are true; later copies are false.

```
%%bpl
(≠v) # v
```

A true/1 count makes one copy; a false/0 count makes none, leaving `22 10 21 5` in their original order.

```
%%bpl
≠ "Banana"
```

The first `B`, `a` and `n` are true; the remaining letters repeat, giving the mask `1 1 1 0 0 0` (`$t $t $t $f $f $f` in BPL). BPL strings use double quotes.

```
%%bpl
3 1 3 1 3 1 # "Banana"
```

Each count says how many copies to make of the corresponding character; `3 1 3 1 3 1` turns `Banana` into `BBBannnannna`.

## ⍣ (Star Diaeresis)

### Dyadic ⍣ (Power Operator)

Power `⍣` repeats a function, feeding each result into the next call. A negative count uses its inverse. Its two operands are the function and the repetition count. [BPL Power reference](https://answerdotai.github.io/basedpl/glyphs/star-diaeresis.html).

```
%%bpl
S ← +↢1
```

BPL's `↢1` fixes the right argument of `+` to `1`, creating the increment function `S` (APL uses `+∘1`).

```
%%bpl
S 0
```

`S` adds one to `0`, returning `1`.

```
%%bpl
(S⍣3) 0
```

`S⍣3` applies increment three times: `0 → 1 → 2 → 3`.

```
%%bpl
add ← {(S⍣⍺) ⍵}
```

`add` increments the right argument `⍵` exactly `⍺` times, adding the left argument to the right.

```
%%bpl
2 add 3
```

Start at `3` and increment twice: `3 → 4 → 5`.

```
%%bpl
mult ← {⍺ (add⍣⍵) 0}
```

`mult` defines multiplication using repeated addition. Take `3 mult 4`:

1. Inside `mult`, `⍺` is `3` and `⍵` is `4`, so the body becomes `3 (add⍣4) 0`.
2. `add⍣4` creates a function that repeats `add` four times. The parentheses group that function.
3. It receives `3` on the left and `0` on the right. The `0` is the starting total. The `⍣` operator repeats `add`, using `3` as its left argument on every call.
4. Each result becomes the next call's right argument:

| Repetition | Call | Result |
|---|---|---|
| 1 | `3 add 0` | `3` |
| 2 | `3 add 3` | `6` |
| 3 | `3 add 6` | `9` |
| 4 | `3 add 9` | `12` |

The final result, `12`, is returned. **4 controls the number of repetitions; 0 supplies the starting total.** The `⍵` in `mult` is the repeat count `4`; each call to `add` receives its own right argument: `0`, `3`, `6`, or `9`.

```
%%bpl
3 mult 4
```

Add `3` four times: `0 → 3 → 6 → 9 → 12`.

```
%%bpl
P ← S⍣¯1
```

`¯1` asks for `S`'s inverse: undo adding one, so `P` subtracts one.

```
%%bpl
P 3
```

`P` decrements `3`, returning `2`.

```
%%bpl
(S⍣¯3) 5
```

A count of `¯3` repeats the inverse three times: `5 → 4 → 3 → 2`.

```
%%bpl
sqr ← *↢2
```

`*↢2` fixes the exponent to `2`, so `sqr` squares its input (APL uses `*∘2`).

```
%%bpl
sqr 3
```

Square `3`: `3 * 2 = 9`.

```
%%bpl
(sqr⍣¯1) 9
```

`sqr⍣¯1` applies the inverse of squaring: the principal square root. It returns `3` for `9`; `¯1` is the repetition count, not an exponent on `9`.

### ⍨ (Tilde Diaeresis): Self

```
%%bpl
pow ← ×⍨
```

```
%%bpl
×⍨ 3
```

With one argument, `⍨` supplies it to both sides of `×`, creating another way to square a number.

```
%%bpl
pow 3
```

`pow 3` means `3 × 3`, giving `9`. The syntax is the same here in APL and BPL.
