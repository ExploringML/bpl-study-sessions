## `/` (Slash)

### Dyadic Function (N-wise Reduce)

````
%%bpl
+/3↕↦4       ⍝ (1+2+3) (2+3+4)
````

````
%%bpl
3÷⍨+/3↕↦4    ⍝ Divide the window sums by 3: moving averages
````

````
%%bpl
+/2↕↦4       ⍝ (1+2) (2+3) (3+4)
````

N-wise reduce sums overlapping groups of N items. `↦4` gives `1 2 3 4`. In these vector examples, `↕` produces a matrix with **one window per row**.

`3↕↦4` gives a **2×3 matrix**:

```text
1 2 3   → 1+2+3 = 6
2 3 4   → 2+3+4 = 9
```

`2↕↦4` gives a **3×2 matrix**:

```text
1 2     → 1+2 = 3
2 3     → 2+3 = 5
3 4     → 3+4 = 7
```

`+/` sums **across each row**, moving horizontally through its columns and collapsing them into one value. Two rows give two sums (`6 9`); three rows give three sums (`3 5 7`). This reduces axis 1. Summing **down each column** instead would use `+⌿`, reducing axis 0.

By default, each window starts one position later. The number of full windows is `input length − window size + 1`: `4−3+1 = 2` or `4−2+1 = 3`. Windows of size three overlap by two items; windows of size two overlap by one.

The middle example divides the sums `6 9` by three with `3÷⍨`, giving moving averages **2 3**.

These adapt Jeremy's APL examples: in BPL, a left argument to `+/` is a starting value (seed), not a window size. Use `↕` for windows, then reduce. [BPL Reduce](https://answerdotai.github.io/basedpl/glyphs/slash.html) · [BPL Windows](https://answerdotai.github.io/basedpl/glyphs/up-down-arrow.html).

````
%%bpl
2↕↦4
````

````
%%bpl
⍝ APL equivalent: 2,/⍳4 (with one-based ⍳)
↓2↕↦4
````

````
%%bpl
⍝ APL equivalent: ¯2,/⍳4 (with one-based ⍳)
↓⌽2↕↦4
````

These examples **keep each pair instead of adding its items**. Jeremy's APL uses comma (`,`) to join the items in each two-item window; plus (`+`) would sum them.

Read the first BPL example from right to left:

1. `↦4` creates `1 2 3 4`.
2. `2↕` makes a matrix whose rows are `1 2`, `2 3`, and `3 4`.
3. `↓` (Split) turns those rows into a list of three separate two-item vectors, matching the nested boxes in the screenshot.

The second example inserts `⌽` before `↓`. On this matrix, `⌽` reverses each row: `1 2` becomes `2 1`, `2 3` becomes `3 2`, and `3 4` becomes `4 3`. The order of the three windows stays the same.

APL's `¯2,/` reverses the order within each window here. **Do not translate that as `¯2↕` in BPL**: a negative window size requests padding. Use the explicit `⌽` reversal instead. [BPL Split](https://answerdotai.github.io/basedpl/glyphs/down-arrow.html) · [BPL Windows](https://answerdotai.github.io/basedpl/glyphs/up-down-arrow.html).

## , (Comma)

### Monadic , (Ravel)

````
%%bpl
⎕ ← cube ← 2 2 2 ⍴ ↦8
````

````
%%bpl
,cube
````

Comma has the same two basic roles in APL and BPL:

- **One input: Ravel.** `,cube` rearranges the cube's entries into one vector, preserving their order: **1 2 3 4 5 6 7 8**. The cube has shape `2 2 2` (two planes, each with two rows and two columns); its ravel has shape `8`. It keeps all eight values rather than summing them.
- **Two inputs: Catenate (concatenate).** `1 2 , 3 4` joins two vectors, giving `1 2 3 4`.

````
%%bpl
⍝ Concatenate: join the lists end to end
1 2,3 4,5 6
````

````
%%bpl
cube cube
````

````
%%bpl
⍴(cube cube)     ⍝ Shape of the two-item list, not of either cube
````

## ⍠ (Axis)

````
%%bpl
3 2 ⍴ ↦6
````

````
%%bpl
1 4 5 =⍠0 (3 2 ⍴ ↦6)
````

This example asks: **does each entry equal the comparison value for its row?** It compares every entry; it does not select rows or reduce the matrix.

1. `↦6` gives `1 2 3 4 5 6`; `3 2 ⍴` arranges them into three rows and two columns.
2. `=⍠0` uses equality with axis 0 specified. The three values `1 4 5` align with the three row positions and are reused for each column.
3. The comparisons are:

| Row | Comparison value | Matrix entries | Results |
|---|---:|---|---|
| First | 1 | `1 2` | True, False |
| Second | 4 | `3 4` | False, True |
| Third | 5 | `5 6` | True, False |

BPL displays True and False as `$t` and `$f`; these correspond to the screenshot's `1` and `0`. The output keeps the matrix's shape, `3 2`.

**APL/BPL difference:** Jeremy writes `=[1]`, using one-based axes in this example. BPL writes `=⍠0`: `⍠` (quad colon) is the Axis operator, and axes always count from zero. Both examples align the vector with the matrix's first axis, the row index. The heading's bracket notation is APL syntax, not BPL syntax. [BPL Axis reference](https://answerdotai.github.io/basedpl/glyphs/quad-colon.html).

````
%%bpl
mat ← 2 3 ⍴ 10 20 30 40 50 60
mat +⍠0 (1 2)     ⍝ Add 1 to row 1 and 2 to row 2
````

`+⍠0` aligns `1 2` with the matrix's first axis (axis 0, the row index). It adds **1 to every entry in the first row** and **2 to every entry in the second row**, giving `11 21 31` and `42 52 62`. Use `⍠` (Quad Colon, Axis), not `⌹` (Domino, Matrix Divide). Jeremy's APL `+[1]` corresponds to BPL `+⍠0` in this example.

````
%%bpl
"ABC"
⍴"ABC"       ⍝ Shape: 3
````

````
%%bpl
1 3 ⍴ "ABC"
⍴(1 3 ⍴ "ABC")       ⍝ Shape: 1 3 (one row, three columns)
````

The letters stay **ABC**; only their shape changes.

- `"ABC"` is a character vector: shape **3**, one axis containing three characters.
- `1 3 ⍴ "ABC"` is a character matrix: shape **1 3**, one row and three columns.

Jeremy's APL `,[0.5]'ABC'` inserts a new length-one axis before the existing character axis. The `0.5` describes where to insert the axis; it does not mean half a character. We use an explicit reshape in BPL to produce the same one-row matrix.

They look almost identical because the matrix has only one row. The extra downward arrow in the display marks its second axis. The `⍴` checks make the distinction visible: `[3]ₓ` versus `[1 3]ₓ`.

````
%%bpl
⎕ ← M ← 2 3 4 ⍴ ↦24
````

````
%%bpl
(,⍠0 1) M             ⍝ Merge planes and rows
⍴((,⍠0 1) M)          ⍝ Shape: 6 4
````

````
%%bpl
(,⍠1 2) M             ⍝ Merge rows and columns
⍴((,⍠1 2) M)          ⍝ Shape: 2 12
````

`M` has shape **2 3 4**: axis 0 selects the two planes, axis 1 the three rows within each plane, and axis 2 the four columns.

Comma with selected axes merges those dimensions into one:

| BPL expression | Axes merged | New shape | Interpretation |
|---|---|---|---|
| `(,⍠0 1) M` | Planes and rows | **6 4**: `(2×3) 4` | Stack the two planes into six rows of four. |
| `(,⍠1 2) M` | Rows and columns | **2 12**: `2 (3×4)` | Read as "ravel axes 1 and 2 together": keep the two planes separate, flattening each plane's 3×4 values into one row of twelve. |

Two selected axes become **one axis whose length is their product**. The first result is `6 4`, not `1 6 4`: no extra length-one axis is inserted. All 24 values stay in the same order. Ordinary `,M` would merge all three axes into a vector of length 24.

Jeremy's APL axes `[1 2]` and `[2 3]` correspond here to BPL `⍠0 1` and `⍠1 2`, because BPL counts axes from zero. [BPL Comma reference](https://answerdotai.github.io/basedpl/glyphs/comma.html).

### Dyadic , (Catenate/Join)

````
%%bpl
1 2 3 , 4 5 6
````

````
%%bpl
cube ← 2 2 2 ⍴ ↦8
cube , 99
````

````
%%bpl
cube (,⍠1) 99
⍴(cube (,⍠1) 99)     ⍝ Shape: 2 3 2
````

Ordinary `cube , 99` joins along the last axis (axis 2), adding a column and changing the shape to **2 2 3**. In `cube (,⍠1) 99`, `⍠1` selects the second axis instead: a new row of `99 99` is added to each plane, giving shape **2 3 2**. The two planes and two columns stay the same. Neither expression changes `cube` unless you assign the result back to it.

````
%%bpl
cube (,⍠0) 99
⍴(cube (,⍠0) 99)     ⍝ Shape: 3 2 2
````

## ⍪ (Comma Bar)

````
%%bpl
⍝ Dyadic comma bar: catenate along the first axis
cube ⍪ 99
⍴(cube ⍪ 99)     ⍝ Shape: 3 2 2
(cube ⍪ 99) ≡ (cube (,⍠0) 99)     ⍝ Match: true
````

We can make 99 appear first by switching the order of comma bar inputs.

````
%%bpl
99 ⍪ cube
⍴(99 ⍪ cube)
````

With two inputs, `⍪` means **catenate first**: it joins along axis 0. You can treat `cube ⍪ 99` as a shortcut for `cube (,⍠0) 99`. Both append a plane filled with `99`, changing the shape from `2 2 2` to `3 2 2`. The `≡` check confirms that the results match.

````
%%bpl
⍝ Monadic comma bar: turn the cube into a matrix
⍪cube
⍴(⍪cube)     ⍝ Shape: 2 4
````

With one input, `⍪` means **table**: make a two-dimensional matrix. It keeps the first axis as the rows and merges the remaining axes into columns. For `cube`, shape `2 2 2` becomes `2 4`: each 2×2 plane becomes one row of four values. This reshapes the data; it does not add values or headings.

````
%%bpl
⍝ Put a header row above two rows of data
headers ← "Name" "Score"
data ← 2 2 ⍴ "Ada" 10 "Ben" 20
headers ⍪ data
⍴(headers ⍪ data)     ⍝ Shape: 3 2
````

This is why comma bar is useful for headings: on a matrix, the first axis runs through the rows. `headers ⍪ data` puts the two header items in a new first row, above the data, giving shape **3 2**. Reversing it to `data ⍪ headers` would put the headings at the bottom. The headings are ordinary array values; `⍪` does not give them special column-name metadata.

[BPL Comma Bar reference](https://answerdotai.github.io/basedpl/glyphs/comma-bar.html).

````
%%bpl
2 3 4 ⍴ $a
````

````
%%bpl
⍪ 2 3 4 ⍴ $a
````

BPL's `$a` replaces APL's `⎕A`. Reshape uses the first 24 letters (A–X), giving shape **2 3 4**. Monadic `⍪` keeps the two planes as rows and merges each plane's 3×4 letters into twelve columns: shape **2 12**.

````
%%bpl
⍝ A vector: one axis with five items
5⍴$a
⍴(5⍴$a)     ⍝ Shape: 5

⍝ Monadic comma bar makes a column matrix
⍪5⍴$a
⍴(⍪5⍴$a)     ⍝ Shape: 5 1
````

````
%%bpl
⍝ Make a column matrix, then transpose it into a row matrix
⍉⍪5⍴$a
⍴(⍉⍪5⍴$a)     ⍝ Shape: 1 5

⍝ The same row matrix can be made directly with reshape
1 5 ⍴ $a
````

Although `⍪5⍴$a` looks like a transpose, the original vector has only one axis: shape **5**. Its horizontal display does not give it a row or column orientation. Monadic `⍪` makes a matrix, keeping the first axis as rows, so this becomes a **5 1 column matrix**.

The same rule turns `2 3 4` into `2 12` and `2 2 2` into `2 4`: keep the first dimension and merge the rest into columns. With a vector, each row contains just one item.

Plain monadic `⍪` on this vector gives `5 1`, not `1 5`. To make a **1 5 row matrix**, use `⍉⍪5⍴$a`: first create the column matrix with `⍪`, then transpose its two axes with `⍉`. Alternatively, reshape directly with `1 5 ⍴ $a`.
