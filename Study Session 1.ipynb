# APL Study Session 1

[BasedPL](https://github.com/AnswerDotAI/basedpl) (BPL) is an array language derived from APL, with ideas from J and BQN. Functions work on whole arrays, and small functions can be combined into larger ones. Its Rust implementation has a Python interface, so it can run inside SolveIt.

This dialogue starts with installation and a few short expressions, then moves on to plots, SVG and canvas animations. The examples are runnable code messages, with notes explaining what to look for.

## `-` (Minus sign;Bar)

### Monadic - (Negate)

```
%bpl - 4
```

```
%bpl - ¯4
```

Note that the negative symbol below is a function applied element-wise to the whole array `-4 2 0 -3 -5`.

```
%bpl -4 2 0 ¯3 ¯5
```

If we had the top negative bar instead as the first symbol then this isn't a function anymore, but just a negative 4 as follows.

```
%bpl ¯4 2 0 ¯3 ¯5
```

### Dyadic - (Minus;subtract)

Dyadic means operations with two arguments, one on the left and on on the right.

```
%bpl 3-2
```

We can substract a fixed value from each array element on the left. It works like broadcasting in NumPy, and PyTorch.

```
%bpl 3 2 1 - 4
```

Or do an element-by-element subtraction operation.

```
%bpl 3 2 1 - 4 5 1
```

```
%bpl 1 - 3 2 1 
```

## `+` (Plus sign)

### Monadic `+` (Conjugate)

Examples one at a time.

```
%bpl + 1.2
```

```
%bpl + 0j4
```

```
%bpl + ¯5j¯6
```

In APL a complex number is represented compactly as `1j2` where the real part is on the left, and imaginary part on the right.

Or examples all at once in one array.

```
%bpl + 1.2 0j4 ¯5j¯6
```

```
%bpl 3j4×+3j4
```

Monadic `+` takes the complex conjugate, so this is `(3+4i)(3-4i)`. The imaginary terms cancel, leaving `3²+4²=25`.

### Dyadic `+` (Plus)

```
%bpl 3+2
```

```
%bpl 3 2 1 + 4
```

```
%bpl 3 2 1 + 4 5 1
```

```
%bpl 1 + 3 2 1 
```
