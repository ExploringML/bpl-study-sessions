## `=` (Equal sign)

### Dyadic `=` (Equal to)

```
%%bpl
1 2 3 = 2 2 3
```

Comparisons in BPL return a distinct Boolean array. BPL displays false and true as `$f` and `$t`, so this result means false, true, true. Traditional APL commonly displays the equivalent Boolean values numerically as `0 1 1`; the comparisons themselves agree.

```
%%bpl
1 2 3 ≡ 2 2 3
```

≡ is Equal Underbar, called Match with two arguments. It compares whole arrays. Here `$f` means the first items differ.

```
%%bpl
1 2 3 ≡ 1 2 3
```

`$t`: both arrays have the same shape and matching items in the same order.

```
%%bpl
1 2 3 ≡ 1 2
```

`$f`: the lengths differ (3 versus 2), so the shapes do not match. ≡ does not broadcast.

```
%%bpl
"banana" = "a"
```

```
%%bpl
"banana" "apple" "candy" = "a"
```

```
%%bpl
7 = "7"
```
